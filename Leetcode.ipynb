Two Sum

```python
nums = [] # ex: 1, 2, 3, 7
nums = input().split()
for i in range(len(nums)):
  nums[i] = int(nums[i])
print(f"Nums: {nums}")
target = int(input()) # target: 9
print(f"Target: {target}")
seen = {} # index : num 0:1 1:2

for i,num in enumerate(nums): #0:1, 1:2 7
  needed = target - num #9-7=2
  if needed in seen:
    print(needed, num)
  else:
    seen[num] = i #seen[2]
```
